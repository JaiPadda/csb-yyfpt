```python
from alphagenome import colab_utils
from alphagenome.data import gene_annotation
from alphagenome.data import genome
from alphagenome.data import transcript as transcript_utils
from alphagenome.interpretation import ism
from alphagenome.models import dna_client
from alphagenome.models import variant_scorers
from alphagenome.visualization import plot_components
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dna_model = dna_client.create(colab_utils.get_api_key())
```

```python
import numpy as np

# =========================================================
# 1. VARIANT (rs642961-style SNP)
# =========================================================
variant = genome.Variant(
    chromosome="chr1",
    position=209815925,   # replace with true rs642961 GRCh38 if needed
    reference_bases="A",
    alternate_bases="G",
)

# =========================================================
# 2. INTERVAL (1MB centered)
# =========================================================
center = variant.position

interval = genome.Interval(
    "chr1",
    center - 500_000,
    center + 500_000
).resize(dna_client.SEQUENCE_LENGTH_1MB)

# =========================================================
# 3. RUN MODEL (ONLY RELIABLE ENHANCER SIGNALS)
# =========================================================
variant_output = dna_model.predict_variant(
    interval=interval,
    variant=variant,
    requested_outputs=[
        dna_client.OutputType.ATAC,          # enhancer accessibility (KEY)
        dna_client.OutputType.CHIP_HISTONE,  # enhancer marks (H3K27ac-like)
        dna_client.OutputType.DNASE,         # accessibility confirmation
        dna_client.OutputType.RNA_SEQ        # downstream proxy effect
        # NOTE: TF removed because it is not consistently available
    ],
    ontology_terms=[
        "UBERON:0002048"  # safe fallback (lung / epithelial proxy)
    ],
)

# =========================================================
# 4. EXTRACT SIGNALS SAFELY
# =========================================================
def get_track(output, name):
    return getattr(output.reference, name).values[:, 0], \
           getattr(output.alternate, name).values[:, 0]

ref_atac, alt_atac = get_track(variant_output, "atac")
ref_dnase, alt_dnase = get_track(variant_output, "dnase")
ref_hist, alt_hist = get_track(variant_output, "chip_histone")
ref_rna, alt_rna = get_track(variant_output, "rna_seq")

# =========================================================
# 5. DIFFERENCE TRACKS
# =========================================================
diff_atac = alt_atac - ref_atac
diff_dnase = alt_dnase - ref_dnase
diff_hist = alt_hist - ref_hist
diff_rna = alt_rna - ref_rna

# =========================================================
# 6. FIND STRONGEST REGULATORY EFFECT (ATAC PRIMARY)
# =========================================================
idx = np.argmax(np.abs(diff_atac))

effect_coord = interval.start + idx
distance_from_variant = effect_coord - variant.position

print("\n=== PRIMARY EFFECT LOCATION ===")
print("Index:", idx)
print("Genomic coord:", effect_coord)
print("Distance from SNP:", distance_from_variant)

# =========================================================
# 7. ENHANCER DISRUPTION SCORING
# =========================================================
def score(diff):
    return np.sum(np.abs(diff))

scores = {
    "ATAC": score(diff_atac),
    "DNase": score(diff_dnase),
    "Histone": score(diff_hist),
    "RNA": score(diff_rna),
}

print("\n=== ENHANCER DISRUPTION SCORES ===")
for k, v in scores.items():
    print(f"{k}: {v:.6f}")

# =========================================================
# 8. BIOLOGICAL INTERPRETATION FLAG
# =========================================================
atac_signal = scores["ATAC"]
hist_signal = scores["Histone"]

if atac_signal > 5 and hist_signal > 5:
    print("\n>>> STRONG EVIDENCE OF ENHANCER DISRUPTION")
elif atac_signal > 1:
    print("\n>>> MODERATE ENHANCER EFFECT")
else:
    print("\n>>> WEAK / NO ENHANCER EFFECT DETECTED")

# =========================================================
# 9. LOCAL ZOOM PLOT (CRITICAL STEP YOU WERE MISSING)
# =========================================================
window = 2000

plt.figure(figsize=(12,4))
plt.plot(ref_atac[idx-window:idx+window], label="REF ATAC")
plt.plot(alt_atac[idx-window:idx+window], label="ALT ATAC")
plt.title("Local enhancer effect (ATAC)")
plt.legend()
plt.show()

# =========================================================
# 10. DIFFERENCE PROFILE (GLOBAL VIEW)
# =========================================================
plt.figure(figsize=(12,3))
plt.plot(diff_atac, label="ALT - REF ATAC")
plt.axvline(idx, linestyle="--", color="red")
plt.title("Allele-specific regulatory change across 1MB")
plt.legend()
plt.show()

# =========================================================
# 11. OPTIONAL: CHECK RNA IMPACT (DOWNSTREAM EFFECT)
# =========================================================
print("\n=== RNA EFFECT SUMMARY ===")
print("Max RNA change:", np.max(np.abs(diff_rna)))
print("Mean RNA change:", np.mean(np.abs(diff_rna)))
```
